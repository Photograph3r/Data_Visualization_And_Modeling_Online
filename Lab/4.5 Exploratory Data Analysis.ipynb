```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'C:\Users\spoll\Desktop\Full Sail\Data Viz\Data_Visualization_And_Modeling_Online\Lab\housing.csv')
df.head()
```

# 4.5 Exploratory Data Analysis Lab

## Dataset Description
**Dataset:** California Housing Data  
**Source:** California Census Data  
**Target Feature:** median_house_value — the median house value for households in a block  

## Features
- **longitude/latitude:** Geographic location of the block
- **housing_median_age:** Median age of houses in the block
- **total_rooms:** Total number of rooms in the block
- **total_bedrooms:** Total number of bedrooms in the block
- **population:** Total population in the block
- **households:** Number of households in the block
- **median_income:** Median income of households (in tens of thousands)
- **ocean_proximity:** How close the block is to the ocean (categorical)

```python
print(df.isnull().sum())
df = df.dropna()
print(f"\nShape after dropping missing values: {df.shape}")
```

```python
sns.scatterplot(data=df, x='median_income', y='median_house_value', alpha=0.3)
plt.title('Median Income vs House Value')
plt.xlabel('Median Income (tens of thousands)')
plt.ylabel('Median House Value ($)')
plt.show()
```

## Visualization 1: Median Income vs House Value
There is a strong positive relationship between median income and house value.
As income increases, house values tend to increase as well. This is the strongest
predictor of house value in the dataset.

```python
sns.boxplot(data=df, x='ocean_proximity', y='median_house_value')
plt.title('Ocean Proximity vs House Value')
plt.xlabel('Ocean Proximity')
plt.ylabel('Median House Value ($)')
plt.xticks(rotation=45)
plt.show()
```

## Visualization 2: Ocean Proximity vs House Value
Houses near the bay and near the ocean have significantly higher median values
than inland houses. Island properties have the highest median value but with 
very few data points. This shows that location relative to the ocean strongly 
influences house prices.

```python
sns.scatterplot(data=df, x='housing_median_age', y='median_house_value', alpha=0.3)
plt.title('Housing Age vs House Value')
plt.xlabel('Housing Median Age')
plt.ylabel('Median House Value ($)')
plt.show()
```

## Visualization 3: Housing Age vs House Value
There is no strong linear relationship between housing age and house value.
Older homes are not necessarily worth less than newer ones, suggesting that
age alone is not a strong predictor of house value in California.

```python
sns.scatterplot(data=df, x='population', y='median_house_value', alpha=0.3)
plt.title('Population vs House Value')
plt.xlabel('Population')
plt.ylabel('Median House Value ($)')
plt.show()
```

## Visualization 4: Population vs House Value
There is no strong relationship between block population and house value.
Population alone does not appear to be a meaningful predictor of housing prices.

```python
sns.histplot(data=df, x='median_income', bins=50)
plt.title('Distribution of Median Income')
plt.xlabel('Median Income')
plt.ylabel('Count')
plt.show()
```

## Visualization 5: Median Income Distribution
Median income is right-skewed, with most households earning between $2,000-$6,000
(in tens of thousands). A small number of very high income blocks pull the 
distribution to the right.

## Big Picture Conclusion
The most important predictors of house value in California are **median income** 
and **ocean proximity**. Blocks with higher incomes and closer proximity to the 
ocean command significantly higher house prices. Features like housing age and 
population have little to no relationship with house value. 

If someone wanted to predict California house prices, they should focus primarily 
on neighborhood income levels and distance to the ocean.
