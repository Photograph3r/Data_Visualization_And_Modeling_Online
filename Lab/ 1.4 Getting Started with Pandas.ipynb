```python
import pandas as pd

df = pd.read_csv(r'C:\Users\spoll\Desktop\Full Sail\Data Viz\Data_Visualization_And_Modeling_Online\Lab\titanic_passengers.csv')
print(df.head())
```

```python
third_class = df[df['Pclass'] == 3]
print(third_class.shape)
```

```python
mean_age = third_class.groupby('Survived')['Age'].mean()
print(mean_age)
```

```python
df['Fare in CAD'] = df['Fare'] * 1.70
print(df[['Fare', 'Fare in CAD']].head(10))
```

```python
crosstab = pd.crosstab(df['Pclass'], df['Survived'])
print(crosstab)
```

```python
port_mapping = {'S': 'Southampton', 'C': 'Cherbourg', 'Q': 'Queenstown'}
df['Port_name'] = df['Embarked'].map(port_mapping)
print(df[['Embarked', 'Port_name']].head(10))
```
