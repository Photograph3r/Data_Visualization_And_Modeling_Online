```python
import pandas as pd

df = pd.read_csv(r'C:\Users\spoll\Desktop\Full Sail\Data Viz\Data_Visualization_And_Modeling_Online\Lab\social media influencers - instagram.csv', encoding='ISO-8859-1')

# Clean up column names
df.columns = df.columns.str.strip()
print(df.head())
```

This dataset contains information on the top 1,000 social media influencers on Instagram. It was collected by Kaggle user Ramjas Maurya and compiled from publicly available Instagram data. The dataset includes features such as the influencer's username, their display name, their primary and secondary content categories, their follower count, their audience's most common country, their authentic engagement, and their average engagement. I chose this dataset because I'm interested in content creation and social media, and I want to understand what separates top influencers from each other especially by category and country.

Which content category has the most influencers?

```python
print(df['category_1'].value_counts().head(10))
```

The most common content category among the top 1,000 Instagram influencers is Music with 235 influencers, followed closely by Cinema & Actors/Actresses with 214. This makes sense as celebrities in entertainment naturally build massive followings. Lifestyle (99) and Sports (113) also rank high, showing that aspirational content dominates the platform.

```python
Which country has the most influencers?
```

```python
print(df['Audience country(mostly)'].value_counts().head(10))
```

The country with the most top Instagram influencers is the United States, with 279 influencers, followed by Brazil (152) and India (149). The US dominance reflects the outsized role American pop culture plays in global social media. It's also notable that Indonesia ranks 4th with 134 influencers, showing that Southeast Asia has a major and growing force in the influencer space.

Who are the top 10 influencers by follower count?

```python
df['Followers_clean'] = df['Followers'].str.replace('M', 'e6').str.replace('K', 'e3').astype(float)
top10 = df.nlargest(10, 'Followers_clean')[['instagram name', 'category_1', 'Followers']]
print(top10)
```

The most followed influencer in this dataset is Instagram's official account with 487.2M followers, followed by Cristiano Ronaldo (419.6M) and Kylie Jenner (323.3M). The top 10 is dominated by athletes, musicians, and actors,  reflecting that traditional celebrity status translates directly into social media reach.

```python
Which category has the highest average engagement?
```

```python
df['Engagement_clean'] = df['Engagement avg'].str.strip().str.replace('M', 'e6').str.replace('K', 'e3').astype(float)
print(df.groupby('category_1')['Engagement_clean'].mean().sort_values(ascending=False).head(10))
```

Despite Music having the most influencers, Management & Marketing has the highest average engagement at approximately 2.05M, followed by Fashion at 2.0M. This suggests that smaller, more niche categories drive stronger audience interaction even with fewer total influencers, quality over quantity.

```python
How many influerncer audiences are primarily based in the US?
```

```python
us_influencers = df[df['Audience country(mostly)'] == 'United States']
print(f"Number of influencers with US audiences: {len(us_influencers)}")
print(us_influencers[['instagram name', 'category_1', 'Followers']].head(10))
```

279 out of 1,000 top Instagram influencers have audiences primarily in the United States, making it by far the most represented country. This reflects the US being the largest English-speaking market on the platform and highlights how American pop culture drives global social media trends.
