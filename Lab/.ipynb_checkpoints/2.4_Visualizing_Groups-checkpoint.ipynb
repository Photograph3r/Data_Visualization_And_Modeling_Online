```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'C:\Users\spoll\Desktop\Full Sail\Data Viz\Data_Visualization_And_Modeling_Online\Lab\titanic_passengers.csv')
df.head()
```

```python
df.groupby('Pclass')['Age'].agg(['mean', 'median'])
```

```python
sns.histplot(data=df[df['Pclass']==1], x='Age', label='Class 1', kde=False)
sns.histplot(data=df[df['Pclass']==2], x='Age', label='Class 2', kde=False)
sns.histplot(data=df[df['Pclass']==3], x='Age', label='Class 3', kde=False)
plt.legend()
plt.title('Age Distribution by Passenger Class')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
df['Pclass_cat'] = df['Pclass'].astype('str')
sns.boxplot(data=df, x='Pclass_cat', y='Age')
plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```

```python
print("""
Objective 5:
- Histogram: Easier to see the shape and overlap of distributions between classes.
- Boxplot: Easier to compare median and IQR precisely between the three classes.

Objective 6:
- Center: Class 1 has the highest median age, Class 3 the lowest.
- Shape: All classes are roughly right-skewed, Class 3 most concentrated in younger ages.
- Spread: Class 1 has the widest spread, Class 3 the most tightly clustered.
""")
```
