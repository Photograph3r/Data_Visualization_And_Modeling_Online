```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r'C:\Users\spoll\Desktop\Full Sail\Data Viz\Data_Visualization_And_Modeling_Online\Lab\spotify-2023.csv', encoding='ISO-8859-1')
print(df.head(10))
```

```python
df['bpm'].plot(kind='hist', bins=30, edgecolor='black', color='mediumslateblue')
plt.title('Distribution of BPM')
plt.xlabel('BPM')
plt.ylabel('Number of Songs')
plt.tight_layout()
plt.show()

print(df['bpm'].describe())
```

BPM
Distribution: Roughly bell-shaped, centered around 120–125 BPM. Mean is 122.5, median is 121.
Engineered hit target: 110–130 BPM

```python
df['key'].value_counts().sort_index().plot(kind='bar', edgecolor='black', color='coral')
plt.title('Distribution of Song Keys')
plt.xlabel('Key')
plt.ylabel('Number of Songs')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(df['key'].value_counts())
```

Key
Distribution: C# is the most common key with 120 songs, followed by G (96) and G# (91).
Engineered hit target: C#

```python
df['mode'].value_counts().plot(kind='bar', edgecolor='black', color='teal')
plt.title('Distribution of Mode (Major vs Minor)')
plt.xlabel('Mode')
plt.ylabel('Number of Songs')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(df['mode'].value_counts())
```

Mode
Distribution: Major dominates with 550 songs vs Minor's 483.
Engineered hit target: Major

```python
df['danceability_%'].plot(kind='hist', bins=30, edgecolor='black', color='hotpink')
plt.title('Distribution of Danceability')
plt.xlabel('Danceability (%)')
plt.ylabel('Number of Songs')
plt.tight_layout()
plt.show()

print(df['danceability_%'].describe())
```

Danceability
Distribution: Left-skewed, clustered between 65–80%. Mean is 67, median is 69.
Engineered hit target: 65–80%

```python
df['energy_%'].plot(kind='hist', bins=30, edgecolor='black', color='orangered')
plt.title('Distribution of Energy')
plt.xlabel('Energy (%)')
plt.ylabel('Number of Songs')
plt.tight_layout()
plt.show()

print(df['energy_%'].describe())
```

Energy
Distribution: Fairly normal, centered around 60–70%. Mean is 64, median is 66.
Engineered hit target: 53–77%

```python
df['speechiness_%'].plot(kind='hist', bins=30, edgecolor='black', color='steelblue')
plt.title('Distribution of Speechiness')
plt.xlabel('Speechiness (%)')
plt.ylabel('Number of Songs')
plt.tight_layout()
plt.show()

print(df['speechiness_%'].describe())
```

Speechiness
Distribution: Heavily right-skewed — most songs cluster near 2–8%. Mean is 10, median is 6.
Engineered hit target: 4–11%

```python
df['acousticness_%'].plot(kind='hist', bins=30, edgecolor='black', color='mediumseagreen')
plt.title('Distribution of Acousticness')
plt.xlabel('Acousticness (%)')
plt.ylabel('Number of Songs')
plt.tight_layout()
plt.show()

print(df['acousticness_%'].describe())
```

Acousticness
Distribution: Heavily right-skewed — most songs are low acousticness, clustered near 0–20%. Mean is 27, median is 18.
Engineered hit target: 6–43%

```python
engineered_hits = df[
    (df['bpm'] >= 110) & (df['bpm'] <= 130) &
    (df['key'] == 'C#') &
    (df['mode'] == 'Major') &
    (df['danceability_%'] >= 65) & (df['danceability_%'] <= 80) &
    (df['energy_%'] >= 53) & (df['energy_%'] <= 77) &
    (df['speechiness_%'] >= 4) & (df['speechiness_%'] <= 11) &
    (df['acousticness_%'] >= 6) & (df['acousticness_%'] <= 43)
]

print(engineered_hits[['track_name', 'artist(s)_name', 'bpm', 'key', 'mode', 
                         'danceability_%', 'energy_%', 'speechiness_%', 'acousticness_%']])
print(f"\nTotal matches: {len(engineered_hits)}")
```

I chose "Cruel Summer" by Taylor Swift as my real smash hit to compare against our engineered hit criteria. Cruel Summer only matched 3 out of 7 of our target features Mode (Major), Energy (72%), and Acousticness (11%). It missed on BPM (170, much faster than our 110–130 target), Key (A instead of C#), Danceability (55%, below our 65–80% range), and Speechiness (15%, above our 4–11% range). This shows that while data can help identify patterns in popular music, a song doesn't need to check every box to be a massive hit. Cruel Summer became one of the biggest songs of the decade despite having a faster tempo and lower danceability than what the data suggests is "typical." Artistry, cultural timing, and artist reputation clearly play a huge role that raw audio features alone can't capture.
