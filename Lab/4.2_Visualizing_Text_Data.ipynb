```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import pandas as pd
import string
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS as sklearn_stop_words

df = pd.read_csv(r'C:\Users\spoll\Desktop\Full Sail\Data Viz\Data_Visualization_And_Modeling_Online\Lab\yelp_subset.csv')
df.head()
```

```python
cleaning_reviews = df[df['categories'].str.contains('cleaning', case=False, na=False)]['text'].str.cat(sep=' ')

cleaning_words = cleaning_reviews.lower().split()
cleaning_words = [word.strip(string.punctuation) for word in cleaning_words 
                  if word.isalnum() 
                  and word not in sklearn_stop_words 
                  and len(word) > 2]

wordcloud = WordCloud(width=800, height=400, background_color='white').generate(' '.join(cleaning_words))
plt.figure(figsize=(10, 7))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Most Frequent Words in Cleaning Business Reviews')
plt.show()
```

```python
auto_reviews = df[df['categories'].str.contains('auto repair', case=False, na=False)]['text'].str.cat(sep=' ')

auto_words = auto_reviews.lower().split()
auto_words = [word.strip(string.punctuation) for word in auto_words 
              if word.isalnum() 
              and word not in sklearn_stop_words 
              and len(word) > 2]

wordcloud2 = WordCloud(width=800, height=400, background_color='white').generate(' '.join(auto_words))
plt.figure(figsize=(10, 7))
plt.imshow(wordcloud2, interpolation='bilinear')
plt.axis('off')
plt.title('Most Frequent Words in Auto Repair Reviews')
plt.show()
```

## Observations
Both word clouds share common words like "service" and "time". 
The cleaning reviews emphasize words like "clean" and "house" while 
auto repair reviews emphasize words like "car" and "oil". 
This shows how word clouds can reveal category-specific language patterns.
